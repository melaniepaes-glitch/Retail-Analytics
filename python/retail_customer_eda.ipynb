# Exploratory Data Analysis (EDA)

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_csv(r"../data/marketing_campaign.csv")
```

## Task 1: General Data Overview

```python
# a. Check the first few rows of the dataset to understand its structure.
print(df.head())
```

```python
# b. Check the data types of each column.
print(df.info())
```

```python
print(df.describe())
```

```python
# c. Check for any missing values in the dataset 
print(df.isna().sum())
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Missing Income:", df["Income"].isna().sum())
```

```python
# Check missing values before cleaning
print("Original records:", len(df))
print("Missing Income values:", df["Income"].isna().sum())

# Remove records with missing Income
df = df.dropna(subset=["Income"]).copy()

# Verify the cleaned dataset
print("Cleaned records:", len(df))
print("Remaining missing Income values:", df["Income"].isna().sum())
```

```python
# Check cleaned dataset
print("Dataset shape:", df.shape)

print("\nMissing values by column:")
print(df.isnull().sum())

print("\nDuplicate customer IDs:")
print(df["ID"].duplicated().sum())

print("\nAverage Income:")
print(round(df["Income"].mean(), 2))
```

```python
# Check for any other missing values
print(df.isnull().sum())
```

## Task 2: Descriptive Statistics

```python
# a. Compute summary statistics for numerical columns (mean, median, min, max, standard deviation).
summary_stats = df.describe()
print("Summary Statistics for Numerical Columns:")
print(summary_stats)
```

```python
# b. Explore the distribution of numerical variables using histograms or box plots.
# Define numerical columns
num_cols = df.select_dtypes(include=np.number).columns
```

```python
# Histograms for numerical variables
for col in num_cols:
    plt.figure(figsize=(8, 6))
    sns.histplot(df[col], kde=True, color='skyblue')
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

## Task 3: Univariate Analysis

```python
# Numerical variables
num_cols = df.select_dtypes(include=np.number).columns
for col in num_cols:
    sns.histplot(df[col], kde=True)
    plt.title(col)
    plt.show()
```

```python
# Box plots for numerical variables
for col in num_cols:
    plt.figure(figsize=(8, 6))
    sns.boxplot(x=df[col], color='lightgreen')
    plt.title(f'Box Plot of {col}')
    plt.xlabel(col)
    plt.show()
```

```python
# Categorical variables
cat_cols = df.select_dtypes(exclude=np.number).columns
for col in cat_cols:
    sns.countplot(data=df, x=col)
    plt.title(col)
    plt.xticks(rotation=45)
    plt.show()
```

## Task 4: Bivariate Analysis

```python
# Numerical vs. Target variable
for col in num_cols:
    sns.boxplot(data=df, x='Response', y=col)
    plt.title(col + ' vs. Response')
    plt.show()
```

```python
# Categorical vs. Target variable
for col in cat_cols:
    sns.countplot(data=df, x=col, hue='Response')
    plt.title(col + ' vs. Response')
    plt.xticks(rotation=45)
    plt.show()
```

```python
# 5. Correlation Analysis
plt.figure(figsize=(16,16))
corr_matrix = df.select_dtypes([np.number]).corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
print("Total customers:", len(df))
print("Customers who responded:", df["Response"].sum())
print("Response rate:", round(df["Response"].mean() * 100, 2), "%")
```

## Exploratory Data Analysis — Key Findings

### 1. Data Quality and Preprocessing
The original dataset contained 2,240 customer records, including 24 records with missing Income values. After removing these records, the cleaned dataset contains 2,216 customers and 30 columns, with no missing values or duplicate customer IDs.

### 2. Customer Income
The average customer income is $52,247.25. The income distribution contains extreme values, so the mean should be interpreted alongside the median and other descriptive statistics.

### 3. Marketing Campaign Response
Of the 2,216 customers, 333 responded positively to the latest marketing campaign, representing a response rate of 15.03%. The remaining 1,883 customers did not respond.

### 4. Business Implications
The relatively low campaign response rate suggests an opportunity to improve customer targeting. Further analysis of income, demographics, purchasing patterns, and previous campaign acceptance may help identify customer segments more likely to respond.

### 5. Limitations
The analysis is exploratory and does not establish causal relationships. Removing customers with missing Income values may introduce selection bias if those customers differ systematically from the retained population.
